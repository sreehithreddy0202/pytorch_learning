ALTERING TENSORS IN PLACE

```python
a=torch.tensor([0,math.pi/4,math.pi/2,(3*math.pi)/4])
print(a)
print(torch.sin(a))
print(a)

b=torch.tensor([0,math.pi/4,math.pi/2,(3*math.pi)/4])
print("\nb:")
print(b)
print(torch.sin_(b))
print(b)
```

```python
a=torch.ones(2,2)
b=torch.rand(2,2)
print(a)
print(b)
print("\n after adding:")
print(a.add_(b))
print(a)
print(b)
print("\n after multiplying:")
print(b.mul_(b))
print(a)
print(b)
```

```python
a=torch.ones(2,2)
b=a
print(a)
print(b)
a[0][1]=560
print(a)
print(b)
a=a*5050
print(a)
print(b)
a[0][1]=56
print(a)
print(b)
```
